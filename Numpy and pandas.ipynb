```python
import pandas as pd
```

```python
data={
    "Name":["Asha","rahul","mappe"],
    "Marks":[98,99,97],
    "Branch":["cse","ece","cil"]
}
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
print(df["Marks"])
```

```python
print(df["Name"])
```

```python
print("Marks row using row")
print(df.iloc[0])
```

```python
print("first person marks",df.iloc[0,1])
```

```python
print("Row slicing of first two rows")
print(df.iloc[0:2])
```

```python
print("Both Rows and coulmns slicing")
print(df.iloc[0:3,0:1])
```

```python
print("filtering")
res=df[df["Marks"]>90]
print(res)
```
