```python
import pandas as pd
import numpy as np
```

```python
data={
    "Student":["A","B","C","D","E","F","G","H","I","J"],

    "Marks":[99,88,89,98,67,78,92,91,97,150]
}

    
```

```python
df=pd.DataFrame(data)
print("Original dataset:")
print(df)
```

```python
Q1=df["Marks"].quantile(0.25)
```

```python
print(Q1)
```

```python
Q3=df["Marks"].quantile(0.75)
```

```python
print(Q3)
```

```python
IQR=Q3-Q1
```

```python
print(IQR)
```

```python
lower_limit=Q1-1.5*IQR
```

```python
upper_limit=Q3+1.5*IQR
```

```python
print("\nQ1:",Q1)
print("\nQ3:",Q3)
print("\nIQR:",IQR)
print("lower limit:",lower_limit)
print("upper limit:",upper_limit)
```

```python
outliers = df[
(df["Marks"]<lower_limit)|(df["Marks"]>upper_limit)
    ]
```

```python
print("Outlier using IQR:")
print(outliers)
```

```python
df["Marks_Treated"] = np.where(df["Marks"] > upper_limit,upper_limit,np.where(df["Marks"]<lower_limit,lower_limit,df["Marks"]))
```

```python
print("----After Treated-----")
print(df)
```

```python
df_capped = df.copy()

df_capped["Marks"] = df_capped["Marks"].clip( 
    lower = lower_limit,
    upper = upper_limit
)
print("After capping")
print(df_capped)
```
